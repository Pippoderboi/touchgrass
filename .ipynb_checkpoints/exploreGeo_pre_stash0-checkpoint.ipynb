```python
import geopandas as gpd
import folium
import pandas as pd
```

```python
gdftoiletten = gpd.read_file("raw_data_geojson/toiletten.geojson")
gdfbaeder = gpd.read_file("raw_data_geojson/baeder.geojson")
gdftheater = gpd.read_file("raw_data_geojson/theater.geojson")
gdfgive_boxen = gpd.read_file("raw_data_geojson/give_boxen.geojson")
gdfgruen = gpd.read_file("raw_data_geojson/gruenflaechen.geojson")
```

```python
#Erste variante mit Pythons len function: count = len(gdftoiletten.index)
#Alternativ mit pandas methode, schließt nullwerte aus: gdftoiletten.index.size
```

```python
datasets = {
    "Toiletten": gdftoiletten,
    "Bäder": gdfbaeder,
    "Theater": gdftheater,
    "Give-Boxen": gdfgive_boxen,
    "Grünflächen": gdfgruen
}

for name, df in datasets.items():
    print(f"{name}: {len(df.index)} Einträge")
```
